# Ejemplo Índice de Masa Corporal (IMC)

```python
altura = [1.7, 1.65, 1.82]
peso = [67, 55, 72]
```

```python
# IMC = Peso / Altura**2
```

```python
np_altura = np.array(altura)
np_peso = np.array(peso)
imc = np_peso / np_altura**2
imc
```

```python
imc > 21
```

```python
imc[imc > 21]
```
