# Práctica: Extracción y Manipulación de Datos con API y Pandas

## Objetivo

El objetivo de esta práctica es que los alumnos aprendan a realizar peticiones GET a una API open source, extraer información relevante y manipularla con pandas para su análisis.

## API Utilizada

Trabajaremos con la API del Art Institute of Chicago, disponible en: https://api.artic.edu/api/v1/artworks

## Actividades

### 1. Instalación de Librerías Necesarias

Asegúrense de tener instaladas las librerías necesarias. Si no las tienen, pueden instalarlas con:

pip install requests pandas

### 2. Realizar una Petición GET a la API

Utilicen requests para obtener información sobre las obras de arte. Extraigan al menos los siguientes campos (La información de los campos los podéis encontrar en: https://api.artic.edu/docs/#collections)


id

title

artist_display

date_display

medium_display

dimensions


### 3. Análisis de los Datos con Pandas

Una vez obtenido el DataFrame, respondan las siguientes preguntas usando pandas:

¿Cuántas obras en total se han obtenido?

¿Cuántas obras tienen información sobre su artista?

¿Cuáles son los 5 artistas con más obras en la base de datos obtenida?
