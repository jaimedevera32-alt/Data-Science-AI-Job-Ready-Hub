## Pandas Cheat Sheet

### Importing Data
- pd.read_csv(filename) | From a CSV file
- pd.read_table(filename) | From a delimited text file (like TSV)
- pd.read_excel(filename) | From an Excel file
- pd.read_sql(query, connection_object) | Read from a SQL table/database
- pd.read_json(json_string) | Read from a JSON formatted string, URL or file.
- pd.read_html(url) | Parses an html URL, string or file and extracts tables to a list of dataframes
- pd.read_clipboard() | Takes the contents of your clipboard and passes it to read_table()
- pd.DataFrame(dict) | From a dict, keys for columns names, values for data as lists

### Exporting Data
- df.to_csv(filename) | Write to a CSV file
- df.to_excel(filename) | Write to an Excel file
- df.to_sql(table_name, connection_object) | Write to a SQL table
- df.to_json(filename) | Write to a file in JSON format

### Viewing/Inspecting Data
- df.head(n) | First n rows of the DataFrame
- df.tail(n) | Last n rows of the DataFrame
- df.shape | Number of rows and columns
- df.info() | Index, Datatype and Memory information
- df.describe() | Summary statistics for numerical columns
- df.isnull().sum() | View null values

### Transform Data
- df.apply(columna) | Transform or create a new column with a custom function
- df.groupby([columns]).agg(aggregatedColumnName=(columnName, typeAgg)) | aggregating the information
- pd.merge(df1, df2, on=column, how=type_join) | Join dataframes

### Selection
- df[col] | Returns column with label col as Series
- df[[col1, col2]] | Returns columns as a new DataFrame
- df.iloc[0,:] | First row
- df.iloc[0,0] | First element of first column

```python
# Crear un DataFrame de ejemplo
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta', 'Sofía'],
    'Edad': [23, 45, 34, 29, 41],
    'Ciudad': ['Madrid', 'Barcelona', 'Sevilla', 'Valencia', 'Bilbao'],
    'Salario': [25000, 32000, 28000, 30000, 27000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Seleccionar una columna
print("\nColumna 'Edad':")
print(df['Edad'])
```

```python
# Filtrar datos
print("\nPersonas con salario mayor a 28000:")
print(df[df['Salario'] > 28000])
```

```python
# Agregar una nueva columna
df['Años de experiencia'] = [1, 10, 7, 5, 8]
print("\nDataFrame con nueva columna:")
print(df)
```

```python
# Operaciones de agregación
print("\nSalario promedio:", df['Salario'].mean())
print("Edad máxima:", df['Edad'].max())
```

```python
# Agrupar datos por ciudad y calcular salario promedio
print("\nSalario promedio por ciudad:")
print(df.groupby('Ciudad')['Salario'].mean())
```

```python
# Ordenar datos por edad
df_sorted = df.sort_values(by='Edad', ascending=True)
print("\nDataFrame ordenado por edad:")
print(df_sorted)
```

```python
# Crear un segundo DataFrame con información adicional
data_extra = {
    'Nombre': ['Ana', 'Luis', 'Pepe', 'Marta', 'Antonio'],
    'Departamento': ['Ventas', 'Marketing', 'TI', 'RRHH', 'Finanzas'],
    'Bonificación': [2000, 2500, 1800, 2200, 2100]
}
df_extra = pd.DataFrame(data_extra)

print("\nSegundo DataFrame:")
df_extra
```

```python
# Unir los DataFrames usando merge
merged_df = pd.merge(df, df_extra , on='Nombre', how='inner')
print("\nDataFrame combinado con merge:")
print(merged_df)
```
