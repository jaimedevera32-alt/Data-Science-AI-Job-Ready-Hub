#**3. Añadir una columna nueva al dataframe user_usage que se llame 'columna_nueva' y cuyo valor sea el doble de 'outgoing_mins_per_month'**

```python
df_user_usage['columna_nueva'] = df_user_usage['outgoing_mins_per_month'] * 2
```

```python
df_user_usage
```

#**4. Añadir una fila nueva al dataframe user_usage con valor = 1 ([1,1,1,1,1,1,1])**

```python
num_columns = len(df_user_usage.columns)
nueva_fila =pd.DataFrame([np.ones(num_columns)], columns=df_user_usage.columns)

df_user_usage = pd.concat([df_user_usage, nueva_fila], ignore_index=True)
```

```python
df_user_usage
```

#**5. Mergear los dataframes user_usage y user_device (columnas = ['use_id', 'platform', 'device']), por clave 'use_id' y utilizando los diferentes métodos de how (inner, outer, left, right) Comentar los resultados**

```python
df_user_usage_device_left = pd.merge(df_user_usage, df_user_device, on='use_id', how='left')
df_user_usage_device_left
```

```python
df_user_usage_device_left.shape
df_user_usage_device_left.isnull().sum()
```

```python
df_user_usage_device_right = pd.merge(df_user_usage, df_user_device, on='use_id', how='right')
df_user_usage_device_right
```

```python
df_user_usage_device_right.shape
df_user_usage_device_right.isnull().sum()
```

```python
df_user_usage_device_inner = pd.merge(df_user_usage, df_user_device, on='use_id', how='inner')
df_user_usage_device_inner
```

```python
df_user_usage_device_inner.shape
df_user_usage_device_inner.isnull().sum()
```

```python
df_user_usage_device_outer = pd.merge(df_user_usage, df_user_device, on='use_id', how='outer')
df_user_usage_device_outer
```

```python
df_user_usage_device_outer.shape
df_user_usage_device_outer.isnull().sum()
```

#**6. Utilizar la función groupby en algunos de los dataframes resultantes anteriores, agrupando por la variable "platform" y haciendo una suma del resto de columnas**

```python
df_user_usage_device_left.groupby(by='platform').sum()
```
