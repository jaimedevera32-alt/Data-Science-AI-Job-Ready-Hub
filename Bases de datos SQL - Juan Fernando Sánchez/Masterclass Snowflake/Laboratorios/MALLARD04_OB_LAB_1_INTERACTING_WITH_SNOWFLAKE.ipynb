## Capacidades del cuaderno de Snowflake 📓 

### Cómo iniciar sesión en un cuaderno

Los cuadernos de Snowflake requieren recursos de cómputo **activos** para ejecutar código en las celdas SQL y Python que contienen. Estos recursos de cómputo (un virtual warehouse de Snowflake, que explicaremos más tarde) son asignados a un Cuaderno de Snowflake en el momento en que se crean. Pero no comienzan en el momento en que ingresa a un cuaderno. Hay dos formas de activar una nueva `session`:

1. Haga clic en el botón **Start*** en la parte superior derecha de la interfaz del cuaderno. Puede tomar varios segundos para iniciar y, una vez iniciado, aparecerá el botón **Active**.
1. Intente ejecutar una celda del cuaderno (SQL o Python) ejecutable. Este sistema establecerá una nueva sesión antes de ejecutar el código en esa celda.

![Iniciar sesión (imagen)](https://edu-cdev-images.s3.us-west-2.amazonaws.com/ob/ob_start_session.png)

💡 **Consejo:** El menú desplegable a la derecha de este botón incluye una opción para terminar (`End`) una sesión. Cuando haya finalizado su trabajo en un cuaderno, se recomienda terminar la sesión de su cuaderno de forma manual. Los cuadernos se cierran automáticamente después de un tiempo de espera de inactividad que puede configurar, pero hacer esto de forma manual una vez finalizado su trabajo le ahorrará costos de cómputo.


### Aspectos básicos de las celdas de cuadernos 📓 

- **Agregar celdas**
    - Desplácese hasta el borde inferior de una celda existente y haga clic en el botón del tipo de celda que desee. 

- **Cambiar tipos de celda**
    - En la celda existente, seleccione la flecha desplegable junto al tipo de celda.

- **Eliminar celdas**
    - Seleccione los puntos suspensivos y elija la acción correspondiente.

- **Duplicar celdas**
    - Seleccione los puntos suspensivos y elija la acción correspondiente.    

- **Mover celdas**
    - Seleccione los puntos suspensivos (tres puntos verticales apilados) y elija la acción correspondiente (Subir celda, Bajar celda).
    - Desplácese sobre la celda que quiere mover, seleccione el ícono de arrastrar y soltar, y mueva la celda a la nueva ubicación.

### Cómo ejecutar celdas de cuadernos 📓 

![Opciones de ejecución del cuaderno (imagen)](https://edu-cdev-images.s3.us-west-2.amazonaws.com/ob/ob_notebook_run_options_1.png)

A continuación se incluyen las acciones habituales con celdas para que las conozca:

- **Ejecutar una celda individual**
    - Seleccione la flecha de reproducción gris en la esquina superior derecha de la celda.
    - Use las teclas de acceso directo de ejecución:
        - Cmd + Return - Mac.
        - Ctrl + Enter - Windows.
    
- **Ejecutar todas las celdas de cuadernos secuencialmente**
    - Seleccione **Run all** (opción arriba del cuaderno a la derecha del botón **Start**).
    - Use teclas de acceso directo:
        - Cmd + Shift + Return – Mac
        - Ctrl + Shift + Enter – Windows

- **Ejecutar celda y avanzar (a la siguiente)**
    - Seleccione los puntos suspensivos arriba de una celda SQL o Python, y haga clic en **Run cell and advance**.
    - Use teclas de acceso directo:
        - Shift + Return – Mac
        - Shift + Enter – Windows

- **Ejecutar todas las anteriores o Ejecutar todas las siguientes (a la celda actual)**
    - Seleccione los puntos suspensivos arriba de una celda SQL o Python, haga clic en **Run all above** o **Run all below**.

💡 Consejo: Tenga en cuenta que el código Python usado para producir el mismo resultado es ligeramente más detallado que la instrucción SQL que ejecutó antes. Se necesitan algunos ajustes más antes de la ejecución, pero es increíblemente flexible. No se preocupe si no comprende los detalles del código en esta etapa. Las explicaremos más detalladamente en este curso.


```
from snowflake.snowpark.context import get_active_session
from snowflake.snowpark.functions import current_date
session = get_active_session()
session.create_dataframe([1]).select(current_date().alias("¿Cuál es la fecha actual?")).collect()
```

## Sustitución de variables de Python en las celdas de cuadernos 🥋

En Cuadernos de Snowflake, podemos usar la sintaxis Jinja `{{..}}` para hacer referencia a las variables de Python definidas antes en celdas Python y SQL que se presentan más adelante.

Demostremos esto definiendo primero una variable Python y tomando una captura de pantalla de la celda. Ejecute la siguiente celda Python para hacerlo:


```
greeting = 'Bienvenido a este curso.'
print(greeting)
```

Podemos volver a usar esta variable almacenada en una celda Python posterior. Ejecute lo siguiente.


```
print("Este es el \"saludo\" que se definió en una celda anterior:")
print(greeting)
```

Podemos usar esta sustitución de variables en celdas **SQL** si aprovechamos la sintaxis de Jinja. Pruebe lo siguiente:


```
SELECT '{{greeting}}' AS "Este es el saludo que se definió en una celda anterior:";
```
