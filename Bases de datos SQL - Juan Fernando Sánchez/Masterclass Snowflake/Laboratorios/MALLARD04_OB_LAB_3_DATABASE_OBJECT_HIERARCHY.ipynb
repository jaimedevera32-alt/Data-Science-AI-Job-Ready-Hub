## Cómo saber qué se ejecutará y qué se ejecutó 📓 

En la celda de un cuaderno, sea Python o SQL, **TODAS** las instrucciones que figuran en una celda se ejecutarán en un bloque cuando se ejecute la celda. Esto significa que el resultado que se muestra estará relacionado con la instrucción final en el bloque si hay varias instrucciones, a menos que se produzca algún error.

El estado de la celda ejecutada se indica por los colores que se muestran en la celda. Este color de estado se muestra en dos lugares: el borde izquierdo de la celda y el mapa de navegación de la celda derecha.

**Color de estado de la celda**:

- **Punto azul**: indica que una celda se modificó, pero aún no se ejecuta.

- **Rojo**: se produjo un error.

- **Verde**: se ejecutó correctamente.

- **Verde en movimiento**: la celda se está ejecutando en este momento.

- **Gris parpadeante**: la celda está a la espera de ejecutarse. Este estado sucede cuando se activan varias celdas para ejecutarse.

**Nota**: Las celdas de Markdown no muestran un estado.

Por ahora, revise y ejecute las siguientes consultas, que proporcionan una muestra de las capacidades SQL de Snowflake en los datos de ejemplo del esquema `TPCH_SF1`.


```
USE SCHEMA snowflake_sample_data.tpch_sf1;

-- recuperar todos los registros de una de las tablas más pequeñas
SELECT * 
FROM NATION;
```

```
-- encontrar las fechas más tempranas y más tardías para los pedidos realizados
SELECT MIN(O_ORDERDATE),
       MAX(O_ORDERDATE) 
FROM ORDERS;
```
