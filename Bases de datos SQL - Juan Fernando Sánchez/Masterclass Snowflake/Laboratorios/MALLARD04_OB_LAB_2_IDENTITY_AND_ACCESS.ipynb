## Bases de datos y esquemas 📓

El conjunto de datos del grupo de bases de datos (tablas y otros objetos). Una agrupación organizacional de segundo nivel dentro de una base de datos se llama **esquema**. Siempre que cree una base de datos, Snowflake creará **automáticamente** dos esquemas para usted.

1. El esquema `INFORMATION_SCHEMA` contiene una colección de vistas. No se puede eliminar (drop) el esquema `INFORMATION_SCHEMA`, cambiar su nombre ni moverlo.

1. El esquema `PUBLIC` se crea vacío y usted puede completarlo con tablas, vistas y otros objetos con el tiempo. Se puede retirar el esquema `PUBLIC`, cambiar su nombre o moverlo en cualquier momento.  


### Ejecute el código para mostrar información sobre los esquemas en la nueva base de datos creada.

El comando `SHOW` también se puede utilizar para ver información general sobre los esquemas incluidos en las bases de datos en Snowflake. Pruebe esto con lo siguiente:


```
SHOW SCHEMAS IN DATABASE {{your_db}};
```
