## Límites y Selección de todo (*) 📓

Una instrucción de seleccionar todo comienza con “`SELECT *`” (que técnicamente es un asterisco). “Select asterisco” es más complicado para expresar. Una selección de todo es una forma de pedir todas las columnas en la tabla sin incluirlas una por una.


### Cómo ejecutar una selección de todo en su nueva tabla 🥋

Vea si puede modificar el código de ejemplo en la siguiente celda SQL para escribir una consulta para seleccionar todos los registros de su nueva tabla `ROOT_DEPTH`.

- Reemplace los caracteres hash (`#`) en la siguiente instrucción con la sintaxis adecuada.


```
-- modifique este código de ejemplo
SELECT * FROM ROOT_DEPTH;
```

### Variaciones de las selecciones de todo 📓

Al trabajar con la selección de todo, Snowflake ofrece algunas funciones SQL convenientes: opciones para ayudarlo a minimizar la escritura necesaria y aún así personalizar el resultado, lo que es muy útil cuando se trabaja con tablas, funciones de tablas y vistas con muchas columnas. Escribir decenas o cientos de nombres de columnas para una consulta no es divertido. Tiene mejores cosas que hacer con su tiempo.

Al especificar `SELECT *`, puede incluir las siguientes opciones: `EXCLUDE`, `REPLACE`, `RENAME` y `ILIKE`.

- `EXCLUDE`: devolverá todas las columnas, excepto aquellas especificadas para su exclusión.
- `REPLACE`: devolverá todas las columnas y reemplazará el valor de las columnas con nombre por el valor de una expresión evaluada.
- `RENAME`: devolverá todas las columnas y especifica los alias de las columnas que deberían usarse en los resultados.
- `ILIKE`: devolverá todas las columnas que coinciden con el patrón especificado.


### Pruebe los ejemplos a continuación en la tabla nueva para personalizar el resultado de la columna con estas variaciones breves.


```
-- devuelva todas las columnas de su tabla, excepto aquellas en la lista de exclusión, utilizando EXCLUDE

SELECT * EXCLUDE (range_min, range_max)
FROM ROOT_DEPTH;
```

```
-- devuelva todas las columnas de su tabla, y agregue la cadena '-RDC' a los valores en la columna root_depth_code, utilizando REPLACE

SELECT * REPLACE (root_depth_code||'-RDC' AS root_depth_code)
FROM ROOT_DEPTH;
```

```
-- devuelva todas las columnas de su tabla, y cambie el título/encabezado de las columnas de rango, utilizando RENAME

SELECT * RENAME (range_min AS root_depth_range_min, 
                 range_max AS root_depth_range_max)
FROM ROOT_DEPTH;
```

```
-- devuelva todas las columnas de su tabla que contengan la cadena `ROOT`, utilizando ILIKE

SELECT * ILIKE '%ROOT%'
FROM ROOT_DEPTH;
```

### Aprender sobre selección de todo y límites 📓  

Si quiere todas las columnas, pero no todas las filas, puede ejecutar una selección de todo con `LIMIT`. Los límites garantizan que tenga apenas un pequeño conjunto de filas. De esa forma, si hay millones de filas, no desperdiciará potencia de cálculo para obtenerlas todas si lo único que quiere es ver algunas.   

Por supuesto, nuestra tabla solo tiene una fila ahora, pero tendrá más filas más adelante. Y más adelante tendremos otras tablas que creamos en las que cargamos muchas más.


```
SELECT *
FROM ROOT_DEPTH
LIMIT 1;
```
