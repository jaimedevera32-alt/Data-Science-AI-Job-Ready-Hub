## ¿Qué es un almacén en Snowflake? 📓 

### Definir “almacén” en Snowflake

- Las personas que han estado trabajando con datos durante un tiempo podrían pensar que el término “almacén de datos” se refiere a una colección especial de estructuras de datos, pero los almacenes(warehouses) no almacenan datos en Snowflake.

- En Snowflake, los almacenes son “recursos de cómputo” que se usan para realizar el procesamiento de los datos. 

- Cuando crea un almacén en Snowflake, usted está definiendo estos “recursos”.

### Escalamiento ascendente y descendente 

- Cambiar el tamaño del almacén cambia la cantidad de servidores en el clúster. 

- Cambiar el tamaño de un almacén actual se conoce como escalamiento ascendente o descendente.

### Escalamiento vertical y horizontal 

- Si está disponible el almacenamiento multiclúster/elástico (Enterprise Edition o versión posterior), un almacén puede escalar horizontalmente en momentos de mayor demanda.

- Si se produce el escalamiento horizontal de multiclúster, los clústeres se agregan por el período de demanda y, luego, se quitan (se contraen) cuando la demanda disminuye.

- La cantidad de servidores en el clúster original determina la cantidad de servidores en cada clúster durante períodos en los que el almacén escala horizontalmente mediante la adición de clústeres. 


## Solo porque puede…📓 

### ...no significa que DEBE 📓 

En Snowflake, usted puede aportar una ENORME potencia de cálculo en tan solo unos pocos segundos. Queremos que sepa que esto es posible, especialmente si tiene un trabajo grande que necesita **GRAN** potencia de cómputo. 

Sin embargo, también queremos que sepa que la mayoría de las consultas **NO** requieren potencia de cómputo masiva. 

De hecho, Snowflake recomienda siempre comenzar con almacenes extra pequeños (XS) y solo hacer el escalamiento ascendente si tiene un motivo convincente para hacerlo. Los almacenes XS usan un crédito por hora. Nuestro almacén más grande es el 6XL optimizado para Snowpark, que usa ¡unos impresionantes 768 créditos por hora! Más de 12 créditos por minuto. 

Para este taller, defina su almacén como XS, excepto en casos en los que le pedimos que use el tamaño S en su lugar. 

Para el uso de Snowflake en el trabajo, probablemente tendrá personas que supervisen la configuración de sus almacenes. El sobredimensionamiento de los almacenes es la forma más simple de cometer errores que provocan grandes sorpresas en la factura mensual, de modo que es mejor acostumbrarse a usar almacenes XS y S la mayor parte del tiempo y escalar verticalmente solo después de una detenida consideración.

Snowflake recomienda que cada cuenta tenga personas que supervisen los costos y que tengan conocimientos avanzados sobre cómo elegir mejor los tamaños de almacén y configurar los ajustes de elasticidad. Estos administradores de costos también podrán calcular si el cambio de tamaño del almacén genera ahorros de tiempo suficientes para justificar los costos incurridos o incluso equilibrarlos.

### Ajustar el tamaño de su almacén virtual mediante código 🥋

Siempre que tenga el permiso para hacer estos tipos de acciones en un almacén virtual, la instrucción `ALTER` puede usarse para cambiar muchas de sus propiedades, cancelar la ejecución de consultas y suspenderlas (detenerlas) o reanudarlas (iniciarlas).

Ejecute la instrucción `ALTER` para regresar su almacén virtual al tamaño que tenía antes del escalamiento vertical usando Snowsight en el ejercicio anterior.


```
ALTER warehouse {{user}}_wh SET warehouse_size=XSMALL;
```

## Configuración de contexto 📓 

El error **Does not exist** aparece con frecuencia y no siempre sucede porque usa el ROL equivocado.

A veces, ve este tipo de error por otros motivos, por ejemplo:

- **Creó** algo **en el lugar equivocado** (como colocar la tabla ROOT_DEPTH en el esquema FRUITS).

- Está **buscando en el lugar equivocado** (como `SELECT * FROM ROOT_DEPTH;`, pero el contexto de su base de datos está configurado en`SNOWFLAKE_SAMPLE_DATA.PUBLIC`).

- Hay un **error de ortografía** (como `SELECT * FROM GARENPLNT.VEGGIES.ROOT_DEPTH;`).

Por supuesto, también es posible que **NO HAYA CREADO** el elemento. Por lo tanto, compruebe también esa posibilidad.

Además, ahora debería saber lo siguiente:

- La diferencia entre un USUARIO y un ROL. 

- La diferencia entre un USUARIO y una CUENTA.

- La diferencia entre ROL PREDETERMINADO, su ROL ACTUAL en la página de inicio de Snowsight y el ROL en el que está definido el contexto de su cuaderno.

- Cómo cambiar de manera programática su rol en un cuaderno de Snowflake.
